Basic action which loads an experiment

---
Use pglActionLoadSession to load a session

```python
# choose session to import from cointreau
a = pglActionLoadSession()
a.configure(dataPath='ssh://gru@cointreau/Users/gru/data')
session = a.run()
```

```python
session.runs[0].display()
```

Fix for missing volume information

```python
# import action from pipeline
from pgl.pglPipeline import pglActionRecreateExperimentDataFromTasks

# configure action
a = pglActionRecreateExperimentDataFromTasks()
a.configure(session=session)
```

```python
fixedSession = a.run()
```

```python
fixedSession.runs[0].display()
```

```python
trialsByParameter = fixedSession.runs[0].getTrialsByParameter('condition')
trialsByParameter.print()
```
