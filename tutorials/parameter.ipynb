Parameters

---

```python
import numpy as np
from pgl.pglParameter import pglParameter, pglParameterBlock, pglParameterNestedBlock, pglParameterBatch
```

Here is a basic parameter that you would use, for example, if you want to have block randomization over orientation between 0 and 180 in 15 deg intervals. The parameter just takes two values, the name and the values that the parameter can take on.

```python
orientationParameter = pglParameter('orientation',np.arange(0,180,15))
```

Once you have initialized it, you can grab values and print them. Note that the way the parameter is returned is as a dict with the parameter name as the key. 

```python
for iTrial in range(24):
    # get the parameter value
    currentParameterValue = orientationParameter.get()
    # and print it
    print(f"{iTrial:2d}: orientation={currentParameterValue['orientation']:3d}")
```

After the parameter has been running you can print to see its state

```python
orientationParameter.print()
```

If you want to ensure the same random sequence, you can set a randomSeed. Try running the below cell multiple times, should give you the same random sequence

```python
# set parameter with the randomSeed
orientationParameter = pglParameter('orientation',np.arange(0,180,15),randomSeed=42)

# print out the values
for iTrial in range(24):
    # get the parameter value
    currentParameterValue = orientationParameter.get()
    # and print it
    print(f"{iTrial:2d}: orientation={currentParameterValue['orientation']:3d}")
```

Add catch trials

---

```python
# The catchTrialEvery parameters makes it so that you will get a catchTrial (parameters set to None) every catchTrialEvery tirals
orientationParameter = pglParameter('orientation',np.arange(0,180,15), catchTrialEvery=8)

# print out the values
for iTrial in range(28):
    # get the parameter value
    currentParameterValue = orientationParameter.get()
    # and print it
    if currentParameterValue['orientation'] == None:
        print(f"{iTrial:2d}: Blank")
    else:
        print(f"{iTrial:2d}: orientation={currentParameterValue['orientation']:3d}")
```

---

Block of two or more parameters

---

```python
# we define two parameters
orientationParameter = pglParameter('orientation',np.arange(0,180,15))
directionParameter = pglParameter('direction',[-1, 1])

# and we want a block to consist of all combinations of the pair
block = pglParameterBlock([orientationParameter, directionParameter])
```

If we display all values for one block, you should be able to confirm that it gives a random ordering of each combination of orientatio and direction

```python
for iTrial in range(12*2):
    # get the parameter values
    currentParameterValues = block.get()
    # and print them
    print(f"{iTrial:2d}: orientation={currentParameterValues['orientation']:3d} direction={currentParameterValues['direction']:2d}")
```

---

Nested block

---

Sometimes what you want is to have all random combinations of the parameters, but you want to go through all values of the first parameter in a block, then through all combinations of the 2nd variable in a bigger block (and so on and so on for more parameters). 

To understand better, try it and see what the difference is from above.

```python
# make the nested block
nestedBlock = pglParameterNestedBlock([orientationParameter, directionParameter])

# and print a block of parameter choices
for iTrial in range(12*2):
    # get the parameter values
    currentParameterValues = nestedBlock.get()
    # and print them
    print(f"{iTrial:2d}: orientation={currentParameterValues['orientation']:3d} direction={currentParameterValues['direction']:2d}")
```

---

batch trials

---

Sometimes you want to grab a few values of the variable at a time, but still have the variables block randomized

```python
# Create 
batch = pglParameterBatch('orientation',np.arange(0,180,15),batchSize=4,catchTrialEvery=8,catchBatchEvery=3)

# and print a block of parameter choices
for iTrial in range(12):
    # get the parameter values
    values = batch.get()
    # and print them
    if values['orientation']:
        print(f"{iTrial:2d}: orientation={' '.join(f'{v:3.0f}' if v is not None else 'None' for v in values['orientation'])}")
    else:
        print(f"{iTrial:2d}: Blank")
```
