---

pRF Experiment

---

```python
# Set up experiment
e = pglExperiment(pgl, settingsName=settingsName, experimentName="pglBarTask", subjectID=subjectID)

# initialize tasks
fixationTask = pglFixationTaskLeftRight(pgl)
barTask = pglBarTask(pgl, randomSeed=42)

# add the tasks to the experiment
e.addTask(barTask)
e.addTask(fixationTask)

# init screen
e.initScreen()

# run experiment
e.run()

# display results
e.display()
```

<div style="background-color:#e6f4ea; padding:15px; border-left:6px solid #2e7d32; color:#1b4332; font-size:18px;">

```python
# initialize experiment
e = pglExperiment(pgl, settingsName=settingsName, experimentName='VWFA', subjectID=subjectID)

# Path to folder containing stimulus .mat files
stimulusFolder = "/Users/justin/proj/C-ShaRP/2026_C-Sharp_VWFA"

# initialize VWFA task
# run 1 → 1-12, run 2 → 13-24, run 3 → 25-36
vwfaTask = pglVWFATask(pgl, stimulusFolder=stimulusFolder, runNum=3, repeatForever=False)

# add the tasks to the experiment
e.addTask(vwfaTask)

# init screen
e.initScreen()

# run experiment
e.run()

# display results
e.display()
```
