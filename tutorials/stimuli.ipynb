Various visual stimuli in pgl

---

```python
# library imports
import PIL.Image as Image

# Load PGL libraries and start a PGL window
from pgl import pgl
pgl = pgl()

# close any existing windows
pgl.cleanUp()

# open a new window
pgl.open(0,800,600,2200,1000)
#pgl.open(0)
pgl.visualAngle(57, 60,40)
```

```python
import numpy as np
# An L-shaped polygon, ordered around its perimeter.
vertices = np.array([
    [-4, -4],
    [ 4, -4],
    [ 4, -1],
    [-1, -1],
    [-1,  4],
    [-4,  4]
])

# CPU-only check.
triangles = pgl.triangulatePolygon(vertices)
assert triangles is not None
assert triangles.shape == (4, 3, 2)

# Finish any previous pass with masking disabled.
pgl.stencilSelect(0)
pgl.flush()

pgl.clearScreen(0)
pgl.polygon(vertices, color=[1, 0, 0])
pgl.flush()
```

```python
# Establish a render-pass boundary before starting stencil creation.
pgl.stencilSelect(0)
pgl.flush()

# Create the circular mask.
assert pgl.stencilCreateBegin(1, invert=True) is not None
pgl.circle(radius=3, fill=True)
assert pgl.stencilCreateEnd() is not None

# Clear before selecting the mask.
pgl.clearScreen([0.1, 0.1, 0.1])

# Draw a rectangle larger than the mask.
assert pgl.stencilSelect(1) is not None
pgl.rect(width=10, height=10, color=[1, 0, 0])
pgl.flush()
```

```python
# Clear the next frame without masking its background quad.
pgl.stencilSelect(0)
pgl.clearScreen([0.1, 0.1, 0.1])

# Reuse the same mask on a second frame.
pgl.stencilSelect(1)
pgl.rect(width=10, height=10, color=[0, 1, 0])
pgl.flush()
```

```python
# Create a circular mask.
assert pgl.stencilCreateBegin(1, invert=True) is not None
pgl.circle(radius=3, fill=True)
assert pgl.stencilCreateEnd() is not None

# Clear before selecting the stencil: clearScreen() draws a background quad.
pgl.clearScreen(0)

# A large red rectangle should appear only inside the circular mask.
assert pgl.stencilSelect(1) is not None
pgl.rect(width=10, height=10, color=[1, 0, 0])
#pgl.circle(radius=3, fill=True)


# Leave subsequent drawing unmasked.
assert pgl.stencilSelect(0) is not None
pgl.flush()
```

Raidal Checkerboard

---

```python
# Init a radial checkerboard stimulus
radialCheckerboard = pgl.radialCheckerboard(pgl,type='sliding',theta=0,innerRadius=0,outerRadius=6,radialWidth=360,checkRadialWidth=7.5,temporalSquareWave=False)

# print the stimulus object
print(radialCheckerboard)

# display duration in second
duration = 2
startTime = pgl.getSecs()

# display the stimulus
while pgl.getSecs() - startTime < duration:
    radialCheckerboard.display()
    pgl.flush()
```

Dots

---

```python
# Init a random dot kinematogram stimulus
rdk = pgl.randomDots(width=10, height=10, noiseType='movshon')

# print the stimulus object
print(rdk)

# display duration in second
duration = 2
startTime = pgl.getSecs()

# display the stimulus
while pgl.getSecs() - startTime < duration:
    rdk.display(direction=181, coherence=1, speed=7.0)
    pgl.flush()
    
```

---

You can align the text in various ways

---

```python
# align text to left
pgl.text("align left",x=1,y=0,xAlign=-1)

# align text to right
pgl.text("align right",x=-1,y=0,xAlign=1)

# align text to top
pgl.text("align top",x=0,y=-1,yAlign=-1)

# align text to bottom
pgl.text("align bottom",x=0,y=1,yAlign=1)

# flush the display
pgl.flush()
```

Instead of an x and y position, you can specify line number to display, where + values are from top of screen and - values are from bottom of screen. Subsequent lines are automatically placed below the last line you draw, and if it goes below the bottom of the screen, it will wrap to the top

```python
# draw a line of text 2 lines from the bottom
pgl.text("Hello again!",line=-2)

# draw two more lines of text, note that the last line will
# wrap to the top of the screen
pgl.text("next line")
pgl.text("and the next line")

# flush the display
pgl.flush()
```

---

Draw a grating. 

---

```python
# initialize the grating
grating = pgl.grating(spatialFrequency=0.1,phase=0, contrast=1,temporalFrequency=4.0,direction=-1)

# clear the screen to gray
pgl.clearScreen((0.5, 0.5, 0.5))
pgl.flush()

# display duration in second
duration = 2
startTime = pgl.getSecs()

# display the grating stimulus
while pgl.getSecs() - startTime < duration:
    grating.display()
    pgl.flush()
```
